What are the key questions we need to answer when analyzing a numeric data set with a single variable?

1. Shape?
2. Skew? Outliers?

The second question may seem like two questions, but skew and outliers occur together for much the same reason.

For **shape**, we hope to find a normal or bell-shaped distribution or to find evidence that the data were, in fact, drawn from such a distribution. We do find uniform distributions often or sometimes bimodal distributions. 

With **skew** and **outliers**, we wish to determine where the outliers, if any exist, lie. To the left? To the right? Are a majority to left or right, or is there a balance? When a majority of outliers lie to one side of the distribution, the distribution is quite likely to be skewed in that direction. We can see this visually as the distribution having a significantly longer tail to one side compared to the other. Where the tail lengthens, there exists more room for outliers as data points exist that are further from the mean and median which often lie quite near the "middle" of the distribution.
